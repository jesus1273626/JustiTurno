### Arreglo de 20 números enteros
En este cuaderno vamos a generar un arreglo con 20 números enteros aleatorios, identificar los números pares e impares, y calcular la suma respectiva de cada grupo.

```python
import numpy as np
import random

# Crear un arreglo de 20 números enteros aleatorios entre 1 y 100 usando NumPy
numeros_lista = [random.randint(1, 100) for _ in range(20)]
numeros = np.array(numeros_lista)
print(f"Arreglo original : {numeros}\n")

# Identificar posiciones y valores pares e impares usando np.where
indices_pares = np.where(numeros % 2 == 0)
indices_impares = np.where(numeros % 2 != 0)

pares = numeros[indices_pares]
impares = numeros[indices_impares]

# Calcular la suma de cada grupo usando np.sum
suma_pares = np.sum(pares)
suma_impares = np.sum(impares)

# Mostrar resultados
print(f"Números pares ({len(pares)}): {pares}")
print(f"Suma de los números pares: {suma_pares}\n")

print(f"Números impares ({len(impares)}): {impares}")
print(f"Suma de los números impares: {suma_impares}")
```

### Identificación de Valores Únicos y Registros Duplicados con NumPy
En esta sección, simularemos una base de datos con números de identificación (IDs) repetidos y utilizaremos NumPy para identificar los registros duplicados.

```python
import numpy as np

ids_base_datos = np.array([101, 102, 103, 101, 104, 105, 102, 106, 101, 107, 105, 108])
print(f"IDs en la base de datos: {ids_base_datos}\n")

valores_unicos, conteos = np.unique(ids_base_datos, return_counts=True)

ids_duplicados = valores_unicos[conteos > 1]

print(f"Valores únicos en la base de datos: {valores_unicos}\n")

print("Análisis de registros duplicados:")
for val, count in zip(valores_unicos, conteos):
    if count > 1:
        indices = np.where(ids_base_datos == val)[0]
        print(f"- El ID {val} está duplicado: aparece {count} veces en las posiciones {indices}")
```

### Registro y Análisis de Asistencia con NumPy
En esta sección, calcularemos las asistencias de 5 aprendices durante 5 días (donde 1 representa asistencia y 0 inasistencia) y determinaremos cuál de ellos obtuvo la mayor asistencia.

```python
import numpy as np

# Definimos los nombres de los 5 aprendices
aprendices = np.array(["Juan", "María", "Carlos", "Ana", "Luis"])

# Simulamos una matriz de asistencia de 5x5 (5 aprendices, 5 días) con valores aleatorios 0 y 1
# Usamos una semilla para asegurar la reproducibilidad del ejercicio
np.random.seed(42)
asistencias_matriz = np.random.randint(0, 2, size=(5, 5))

print("Matriz de Asistencia (Filas: Aprendices, Columnas: Días 1-5):")
print(asistencias_matriz)
print()

# Calcular el total de asistencias de cada aprendiz sumando por filas (axis=1)
total_asistencias = np.sum(asistencias_matriz, axis=1)

# Mostrar el total de cada uno
print("Total de asistencias por aprendiz:")
for i, nombre in enumerate(aprendices):
    print(f"- {nombre}: {total_asistencias[i]} días asistidos")
print()

# Determinar quién tuvo la mayor asistencia
indice_maximo = np.argmax(total_asistencias)
max_asistencias = total_asistencias[indice_maximo]

# Podría haber un empate, así que buscamos todos los que tengan el puntaje máximo
mejores_aprendices = aprendices[total_asistencias == max_asistencias]

print(f"El nivel de asistencia más alto fue de: {max_asistencias} días")
print(f"Aprendiz(ces) con mayor asistencia: {', '.join(mejores_aprendices)}")
```

### Ejercicio 2. Estadística descriptiva con NumPy

```python
# Convertir columnas de Pandas en arreglos NumPy
edades = df['Age at enrollment'].to_numpy()
notas_admision = df['Admission grade'].to_numpy()

# Estadísticas de edad
print("--- ESTADÍSTICAS DE EDAD (NumPy) ---")
promedio_edad = np.mean(edades)
mediana_edad = np.median(edades)
min_edad = np.min(edades)
max_edad = np.max(edades)
desviacion_edad = np.std(edades)

print(f"Edad promedio: {promedio_edad:.2f} años")
print(f"Mediana de edad: {mediana_edad:.1f} años")
print(f"Edad mínima: {min_edad} años")
print(f"Edad máxima: {max_edad} años")
print(f"Desviación estándar: {desviacion_edad:.2f} años")

# Filtrado de estudiantes por rango de edad usando condicionales de NumPy
mayores_25 = np.sum(edades > 25)
menores_20 = np.sum(edades < 20)
print(f"Estudiantes mayores de 25 años: {mayores_25}")
print(f"Estudiantes menores de 20 años: {menores_20}")

# Estadísticas de notas de admisión
print("\n--- ESTADÍSTICAS DE NOTAS DE ADMISIÓN (NumPy) ---")
promedio_nota = np.mean(notas_admision)
desviacion_nota = np.std(notas_admision)
print(f"Promedio de calificación de admisión: {promedio_nota:.2f}")
print(f"Desviación estándar de calificación de admisión: {desviacion_nota:.2f}")
```

   La media (23.27 años) es mayor que la mediana (20.0 años). Esto indica que la distribución de las edades tiene un sesgo hacia la derecha (sesgo positivo); la gran mayoría de los estudiantes son jóvenes (cercanos a 18-20 años), pero hay un grupo menor de estudiantes de mayor edad (hasta 70 años) que eleva el promedio estadístico.
2. **¿Es la edad una variable homogénea o heterogénea?**
   Es relativamente heterogénea. Aunque hay una concentración de estudiantes de 18 a 20 años, la desviación estándar de ~6.33 años y el rango amplio (de 17 a 70 años) muestran una diversidad importante en el aula.

### Ejercicio 4. Relación entre condiciones financieras y académicas

```python
# 1. Tabla cruzada de matrícula al día frente a deserción académica
tabla_matricula = pd.crosstab(df['Tuition fees up to date'], df['Target'], margins=True)
print("--- TABLA DE MATRÍCULA AL DÍA VS TARGET ---")
display(tabla_matricula)

# Expresar en porcentajes por fila para análisis comparativo
tabla_matricula_pct = pd.crosstab(df['Tuition fees up to date'], df['Target'], normalize='index') * 100
print("\n--- TABLA EN PORCENTAJES (%) POR FILA ---")
display(tabla_matricula_pct.round(2))

# 2. Tabla de contingencia de becados frente a la variable Target
tabla_becados = pd.crosstab(df['Scholarship holder'], df['Target'], margins=True)
tabla_becados_pct = pd.crosstab(df['Scholarship holder'], df['Target'], normalize='index') * 100

print("\n--- TABLA DE BECADOS VS TARGET ---")
display(tabla_becados)
print("\n--- TABLA DE BECADOS EN PORCENTAJES (%) ---")
display(tabla_becados_pct.round(2))
```

### Ejercicio 5. Análisis del rendimiento académico primer vs segundo semestre

```python
# 1. Promedio general de calificaciones en el primer y segundo semestre
promedio_sem1 = df['Curricular units 1st sem (grade)'].mean()
promedio_sem2 = df['Curricular units 2nd sem (grade)'].mean()

print("--- RENDIMIENTO ACADÉMICO PROMEDIO ---")
print(f"Calificación promedio global - Semestre 1: {promedio_sem1:.2f}")
print(f"Calificación promedio global - Semestre 2: {promedio_sem2:.2f}")

# 2. Coeficiente de correlación de Pearson
correlacion = df['Curricular units 1st sem (grade)'].corr(df['Curricular units 2nd sem (grade)'])
print(f"\nCoeficiente de correlación de Pearson entre semestres: {correlacion:.4f}")

# 3. Comparación de promedios de calificación divididos por Target
promedio_por_target_s1 = df.groupby('Target')['Curricular units 1st sem (grade)'].mean()
promedio_por_target_s2 = df.groupby('Target')['Curricular units 2nd sem (grade)'].mean()

print("\n--- PROMEDIO SEMESTRE 1 POR TARGET ---")
for cat, val in promedio_por_target_s1.items():
    print(f"- {cat}: {val:.2f}")

print("\n--- PROMEDIO SEMESTRE 2 POR TARGET ---")
for cat, val in promedio_por_target_s2.items():
    print(f"- {cat}: {val:.2f}")
```

## Mini-Informe de Hallazgos y Recomendaciones (Ejercicio 7)

### 1. Introducción
Este análisis utiliza un conjunto de datos académicos de **4,424 estudiantes** con 37 variables sociodemográficas y de rendimiento para comprender qué factores se asocian con el éxito educativo (`Graduate`) y la deserción (`Dropout`).

### 2. Cinco (5) Hallazgos Clave del Análisis
1. **Factor Financiero Decisivo:** Los estudiantes con matrículas pendientes tienen un comportamiento alarmante; el **89.96%** de ellos termina en deserción (`Dropout`). El aspecto financiero es el factor de riesgo individual más crítico.
2. **La Beca como Factor de Retención:** Los estudiantes becados (`Scholarship holder = 1`) tienen una excelente tasa de graduación del **75.46%** y apenas un 10.15% de deserción, demostrando la alta efectividad del apoyo financiero.
3. **Edad de Ingreso:** Los estudiantes que desertan ingresan con una edad significativamente más alta en promedio (**24.87 años**) en comparación con los que logran graduarse (**21.88 años**).
4. **Desplome del Desempeño Académico:** El rendimiento en el primer semestre correlaciona de manera muy fuerte con el segundo (coeficiente de Pearson = **0.8372**). Los desertores muestran una caída severa de su promedio de calificaciones, que pasa de 7.26 a 5.90.
5. **Bajo Impacto de Variables Macro:** Los factores económicos globales como la tasa de desempleo o inflación del entorno muestran correlaciones lineales insignificantes (casi nulas, menores a 0.1) con el rendimiento académico directo de los estudiantes.

1. **Sistema de Alerta Temprana en el Primer Semestre:** Implementar tutorías académicas obligatorias de apoyo inmediato para aquellos alumnos cuyo promedio del primer semestre caiga por debajo de 10 puntos, previniendo el desplome del segundo semestre.
